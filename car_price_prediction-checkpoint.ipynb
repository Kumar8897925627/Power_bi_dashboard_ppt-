```python
df = df.dropna()
```

```python
df['Mileage'] = df['Mileage'].fillna(df['Mileage'].median())
print(df)
df['Engine Size'] = df['Engine Size'].fillna(df['Engine Size'].mean())
print(df)
df['Price'] = df['Price'].fillna(df['Price'].median())
print(df)
```

```python
df['Fuel Type'] = df['Fuel Type'].fillna(df['Fuel Type'].mode()[0])
print(df['Fuel Type'])
df['Transmission'] = df['Transmission'].fillna(df['Transmission'].mode()[0])
print(df['Transmission'])
```

```python
df['Year'] = df['Year'].astype(int)
print(df)
df['Mileage'] = df['Mileage'].astype(int)
print(df)
df['Engine Size'] = df['Engine Size'].astype(float)
print(df)
df['Price'] = df['Price'].astype(float)
print(df)
```

```python
df['Brand'] = df['Brand'].str.strip()
df['Fuel Type'] = df['Fuel Type'].str.strip()
df['Transmission'] = df['Transmission'].str.strip()
df['Condition'] = df['Condition'].str.strip()
df['Model'] = df['Model'].str.strip()
print(df)
```

```python
df['Brand'] = df['Brand'].str.upper()
df['Transmission'] = df['Transmission'].str.title()
df['Fuel Type'] = df['Fuel Type'].str.title()
df['Condition'] = df['Condition'].str.replace('_', ' ').str.title()
(df)
```

```python
df['Fuel Type'].replace({'petrol':'Petrol', 'diesel':'Diesel', 'electric':'Electric'})
df['Transmission'].replace({'Manual gearbox':'Manual', 'Auto':'Automatic'})
df
```

```python
df['Car_Age'] = 2025 - df['Year']
df
```
